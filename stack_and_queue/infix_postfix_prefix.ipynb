```python
def pos_inf(s):
    st = []
    i = 0
    while i < len(s):
        if (s[i] <='Z' and s[i] >= 'A') or (s[i] <='z' and s[i] >= 'a') or \
        s[i].isdigit():
            st.append(s[i])
        else:
            op1 = st.pop()
            op2  = st.pop()
            expr = '(' + op2 + s[i] + op1 + ')'
            st.append(expr)
        i+=1
    return st[-1]
t = "pq+mn-*"
pos_inf(t)
```

```python
def pos_to_inf(s):
    st = []
    for ch in s:
        if ch.isalpha() or ch.isdigit():
            st.append(ch)
        else:
            op1 = st.pop()
            op2 = st.pop()
            exp = '(' + op2 + ch + op1 + ')'
            st.append(exp)
    return st[-1]

t = "pq+mn-*"
pos_to_inf(t)
```

```python
def pre_to_in(s):
    s = s[::-1]
    st = []
    for ch in s:
        if ch.isalpha() or ch.isdigit():
            st.append(ch)
        else:
            op1 = st.pop()
            op2 = st.pop()
            exp = '(' + op1 + ch + op2 + ')'
            st.append(exp)
    return st[-1]
s = "+-+a*b^cd*^efgh"
pre_to_in(s)
```
