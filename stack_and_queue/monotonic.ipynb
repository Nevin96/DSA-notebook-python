```python
# optimised
def nxt_i(a):
    nge = [-1]*len(a)
    st = []
    for i in range(len(a)-1,-1,-1):
        while st and st[-1] <= a[i]:
            st.pop()
        else:
            nge[i] = st[-1]
        st.append(a[i])
    return nge
a = [6,0,8,1,3]
nxt_i(a)
```

```python
def next_gr(a,b):
    st = []
    nge_map = {}
    for num in b[::-1]:
        while st and num >= st[-1]:
            st.pop()
        if st:
            nge_map[num] = st[-1]
        else:
            nge_map[num] = -1
        st.append(num)
    return [nge_map[num] for num in a]

nums1 = [4,1,2]
nums2 = [1,3,4,2]
next_gr(nums1,nums2)
```

```python
# brute
def nxt_grr(a):
    nge = []
    for i in range(len(a)):
        for j in range(i+1,len(a)):
            if a[j] > a[i]:
                nge.append(a[j])
                break
        else:
            for j in range(0,i+1):
                if a[j] > a[i]:
                    nge.append(a[j])
                    break
            else:
                nge.append(-1)
        
    return nge

nums = [1,2,1]
nxt_grr(nums)
```

```python
# optimal
def nxt_gr_2(a):
    nge = [-1] * len(a)
    st = []
    for i in range(2*len(a)-1,-1,-1):
        while st and st[-1] <= a[i%len(a)]:
            st.pop()
        if st:
            nge[i%len(a)] = st[-1]
        st.append(a[i%len(a)])
    return nge

nums = [1,2,1]
nxt_gr_2(nums)
```

```python
# optimal
def gr_el(a,ind):
    n = len(a)
    st = []
    counts = [0] * n
    for i in range(n-1,-1,-1):
        while st and a[st[-1]] <= a[i] :
            st.pop()
        if st:
            counts[i] = 1 + counts[st[-1]]
        st.append(i)
    return [counts[i] for i in ind]

arr = [1, 2, 3, 4, 1]
indices = [0, 3]
gr_el(arr,indices)
```

```python
# brute 
def tap_rain_water(a):
    n = len(a)
    def prefix_s(a):
        prefix = [0] * n
        prefix[0] = a[0]
        for i in range(1,len(a)):
            prefix[i] = max(a[i],prefix[i-1])
        return prefix
    def suffix_s(a):
        suffix = [0]*n
        suffix[n-1] = a[n-1]
        for i in range(n-2,-1,-1):
            suffix[i] = max(a[i],suffix[i+1])
        return suffix
    left_max = prefix_s(a)
    right_max = suffix_s(a)
    total = 0
    for i in range(n):
        total += min(left_max[i],right_max[i]) - a[i]
    return total
b =[0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
tap_rain_water(b)
```

```python
# optimal
def tap_r(a):
    n = len(a)
    lmax,rmax,total = 0,0,0
    l = 0
    r = n-1
    while l < r:
        if a[l] <= a[r]:
            if lmax > a[l]:
                total += lmax - a[l]
            else:
                lmax = a[l]
            l += 1
        else:
            if rmax > a[r]:
                total += rmax - a[r]
            else:
                rmax = a[r]
            r -=1
    return total

b =[0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
tap_r(b)
```

```python
# brute
def minimum_sum_sbaray(a):
    n = len(a)
    sum =0
    for i in range(n):
        mini = a[i]
        for j in range(i,n):
            if mini > min(a[j],a[i]):
                mini = min(a[j],a[i])
            sum += mini
    return sum
a = [11,81,94,43,3]
minimum_sum_sbaray(a)
```

```python
# optimal
def sum_sub(a):
    n = len(a)
    def nse(a):
        st = []
        arr = [n] * n
        for i in range(n-1,-1,-1):
            while st and a[st[-1]] >= a[i]:
                st.pop()
            if st:
                arr[i] = st[-1]
            st.append(i)
        return arr
    def pse(a):
        st = []
        arr = [-1] * n
        for i in range(n):
            while st and a[st[-1]] > a[i]:
                st.pop()
            if st:
                arr[i] = st[-1]
            st.append(i)
        return arr
    q = pse(a)
    w = nse(a)
    total = 0
    for i in range(n):
        left = i - q[i]
        right = w[i] - i
        total += left*right*a[i]
    return total

c = [3,1,2,4]
sum_sub(c)
```

```python
def col(a):
    n = len(a)
    st = []
    for i in range(n):
        if a[i] > 0:
            st.append(a[i])
        else:
            while st and st[-1] > 0 and st[-1] < abs(a[i]):
                st.pop()
            if st and st[-1] == abs(a[i]):
                st.pop()
            elif not st or st[-1] < 0:
                st.append(a[i])
    return st

b = [2,-2]
col(b)
```

```python
# brute
def su(a):
    n = len(a)
    total = 0
    for i in range(n):
        mini = a[i]
        maxi = a[i]
        for j in range(i,n):
            mini = min(mini,min(a[i],a[j]))
            maxi = max(maxi,max(a[i],a[j]))
            total += maxi - mini
    return total
b = [4,-2,-3,4,1]
su(b)
```
