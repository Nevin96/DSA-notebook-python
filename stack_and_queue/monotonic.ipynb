```python
def nxt_great(a,b):
    c = []
    for i in range(len(a)):
        for j in range(len(b)):
            if a[i] ==  b[j]:
                for g in range(j,len(b)):
                    if b[g] > a[i]:
                        c.append(b[g])
                        break
                else:
                    c.append(-1)
    return c
nums1 = [4,1,2]
nums2 = [1,3,4,2]
nxt_great(nums1,nums2)
```

```python
# bruteforce
def nxt(a):
    c = []
    for i in range(len(a)):
        for j in range(i+1,len(a)):
            if a[j] > a[i]:
                c.append(a[j])
                break
        else:
            c.append(-1)
    return c
a = [6,0,8,1,3]
nxt(a)
```

```python
# optimised
def nxt_i(a):
    nge = [-1]*len(a)
    st = []
    for i in range(len(a)-1,-1,-1):
        while st and st[-1] <= a[i]:
            st.pop()
        else:
            nge[i] = st[-1]
        st.append(a[i])
    return nge
a = [6,0,8,1,3]
nxt_i(a)
```

```python
def nxt_gre(a):
    nge = [-1] * len(a)
    st = []
    for i in range(len(a)-1,-1,-1):
        while st and a[i] >= st[-1]:
            st.pop()
        if st:
            nge[i] = st[-1]
        st.append(a[i])
    return nge
a = [6,0,8,1,3]
nxt_gre(a)
```

```python
def next_gr(a,b):
    st = []
    nge_map = {}
    for num in b[::-1]:
        while st and num >= st[-1]:
            st.pop()
        if st:
            nge_map[num] = st[-1]
        else:
            nge_map[num] = -1
        st.append(num)
    return [nge_map[num] for num in a]

nums1 = [4,1,2]
nums2 = [1,3,4,2]
next_gr(nums1,nums2)
```

```python
# brute
def nxt_grr(a):
    nge = []
    for i in range(len(a)):
        for j in range(i+1,len(a)):
            if a[j] > a[i]:
                nge.append(a[j])
                break
        else:
            for j in range(0,i+1):
                if a[j] > a[i]:
                    nge.append(a[j])
                    break
            else:
                nge.append(-1)
        
    return nge

nums = [1,2,1]
nxt_grr(nums)
```

```python
# optimal
def nxt_gr_2(a):
    nge = [-1] * len(a)
    st = []
    for i in range(2*len(a)-1,-1,-1):
        while st and st[-1] <= a[i%len(a)]:
            st.pop()
        if st:
            nge[i%len(a)] = st[-1]
        st.append(a[i%len(a)])
    return nge

nums = [1,2,1]
nxt_gr_2(nums)
```

```python
def sm(a):
    n = len(a)
    st = []
    nge = [-1] * n
    for i in range(n-1,-1,-1):
        while st and st[-1] >= a[i]:
            st.pop()
        if st:
            nge[i] = st[-1]
        st.append(a[i])
    return nge

arr = [4, 8, 5, 2, 25]
sm(arr)
```
