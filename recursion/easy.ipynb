```python
def su(a,ind = 0,b =None):
    if b  == None:
        b = []
    n = len(a)
    if ind >= n:
        print(b)
        return 
    b.append(a[ind])
    su(a,ind+1,b)
    b.pop()
    su(a,ind+1,b)
a = [3,1,2]
su(a)
```
