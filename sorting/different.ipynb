```python
def merge_sort(a):
    n = len(a)
    low = 0
    high = n-1
    def merge(a,low,mid,high):
        left = low
        right = mid+1
        b = []
        while left <= mid and right <= high:
            if a[left] < a[right]:
                b.append(a[left])
                left+=1
            else:
                b.append(a[right])
                right+=1
        while left <= mid:
            b.append(a[left])
            left +=1
        while right <= high:
            b.append(a[right])
            right+=1
        for i in range(len(b)):
            a[low+i] = b[i]
    def divide(a,low,high):
        if low >= high:
            return
        mid = (low+high)//2
        divide(a,low,mid)
        divide(a,mid+1,high)
        merge(a,low,mid,high)
    divide(a,low,high)
    return a
a= [9,4,2,1,8,9]
merge_sort(a)
```

```python
def insertion_sort(a):
    n = len(a)
    for i in range(1,n):
        j = i
        while j > 0 and a[j-1] > a[j]:
            a[j-1],a[j] = a[j],a[j-1]
            j-=1
    return a
a= [9,4,2,1,8,9]
insertion_sort(a)
```

```python
def selection_sort(a):
    n = len(a)
    for i in range(n):
        mini = i
        for j in range(i+1,n):
            if a[j] < a[mini]:
                mini = j
        a[i],a[mini] =a[mini],a[i]
    return a
b= [9,4,2,1,8,9]
selection_sort(b)
```

```python
def bubble_sort(a):
    n = len(a)
    for i in range(n-1,-1,-1):
        swap = False
        for j in range(i):
            if a[j] > a[j+1]:
                a[j],a[j+1] = a[j+1],a[j]
                swap = True
        if swap == False:
            break
    return a
b= [9,4,2,1,8,9]
bubble_sort(b)
```

```python
def quick_sort(a):
    n = len(a)
    low= 0
    high = n-1
    def piv(a,low,high):
        pivot = a[low]
        i = low
        j = high
        while i < j:
            while a[i] <= pivot and i <= high-1:
                i+=1
            while a[j] > pivot and j >= low-1:
                j-=1
            if i < j:
                a[i],a[j] = a[j],a[i]
        a[low],a[j] = a[j],a[low]
        return j
    def pi(a,low,high):
        if low < high:
            pivot = piv(a,low,high)
            pi(a,low,pivot-1)
            pi(a,pivot+1,high)
    pi(a,low,high)
    return a
b= [9,4,2,1,8,9]
quick_sort(b)
```
