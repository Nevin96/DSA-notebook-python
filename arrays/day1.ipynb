```python
def lar_2(a):
    n = len(a)
    maxi_2 = float('-inf')
    maxi = max(a)
    for i in range(n):
        if a[i] < maxi and a[i] > maxi_2:
            maxi_2 = a[i]
    if maxi_2 == float('-inf'):
        return -1
    return maxi_2
b =    [10, 10, 10, 10, 10]
lar_2(b)
```

```python
def srt(a):
    n = len(a)
    for i in range(n-1):
        if a[i] > a[i+1]:
            return False
    return True
nums = [1, 2, 1, 4, 5]
srt(nums)
```

```python
def dup(a):
    n = len(a)
    j = 0
    for i in range(n):
        if a[i] != a[j]:
            a[j+1] = a[i]
            j+=1
    return j+1
            
        
nums = [-2, 2, 4, 4, 4, 4, 5, 5]
dup(nums)
```

```python
def rot(a,k):
    n = len(a)
    k = k % n
    if n < 2:
        return a
    def rev(start,end):
        while start < end:
            a[start],a[end] = a[end],a[start]
            start += 1
            end -=1
    rev(0,n-1)
    rev(k,n-1)
    rev(0,k-1)
    return a
nums = [1, 2, 3, 4, 5, 6]
rot(nums,2)
```

```python
# brute
def move_0(a):
    n = len(a)
    c = []
    for i in range(n):
        if a[i] != 0:
            c.append(a[i])
    for i in range(len(c)):
        a[i] = c[i]
    for i in range(len(c),n):
        a[i] = 0 
    return a
b = [0, 1, 4, 0, 5, 2]
move_0(b)
```

```python
def mov_0_be(a):
    n = len(a)
    j = -1
    for i in range(n):
        if a[i] == 0:
            j = i
            break
    if j != -1:
        for q in range(j+1,n):
            if a[q] != 0:
                a[j],a[q] = a[q],a[j]
                j+=1
    return a
b = [8, 1, 0, 0, 5, 2]
mov_0_be(b)
```

```python
def longest(nums):    
    a =  set(nums)
    longest = 0
    for i in a:
        if i-1 not in nums:
            length = 1
            current = i
            while current+1 in a:
                length+=1
                current +=1
            longest = max(longest,length)
    return longest
c = [100,4,200,1,3,2]
longest(c)
```

```python
def two_sum(a,k):
    g = {}
    for i in range(len(a)):
        comp = k- a[i]
        if comp in g:
            ind = g[comp]
            return [ind,i]
        if a[i] not in g:
            g[a[i]] = i

nums = [2,7,11,5]
target = 9
two_sum(nums,target)
```

```python
def miss_num(a):
    n = len(a)
    s = sum(a)
    su = (n*(n+1))/2
    return su- s

nums = [0, 1, 2, 4, 5, 6]
miss_num(nums)
```

```python
def max_cons_ones(a):
    count = 0
    maxi = 0
    for i in a:
        if i == 1:
            count+=1
        else:
            count = 0
        maxi = max(maxi,count)
    return maxi

b = [1, 1, 0, 0, 1, 1, 1, 0]
max_cons_ones(b)
```

```python
def single(a):
    g = {}
    for i in a:
        if i in g:
            g[i] += 1
        else:
            g[i] = 1
    for k,v in g.items():
        if v == 1:
            return k
nums = [5]
single(nums)
```
