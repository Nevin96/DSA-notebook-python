```python
def ncr_pascal(n,r):
    res = 1
    for i in range(r):
        res = res*(n-i)
        res = res / (i+1)
    return res

ncr_pascal(3,2)
```

```python
def el_in_row(n):
    ans = []
    for i in range(n):
        c = ncr_pascal(n-1,i)
        ans.append(c)
    return ans
el_in_row(3)
```

```python
def el_in_pas(n):
    ans = []
    res = 1
    ans.append(res)
    for i in range(1,n):
        res = (res * (n - i))//i
        ans.append(res)
    return ans
el_in_pas(3)
```

```python
def pascal_tr(n):
    ans = []
    for i in range(1,n+1):
        l = []
        for j in range(i):
            c = ncr_pascal(i-1,j)
            l.append(c)
        ans.append(l)
    return ans
pascal_tr(3)
```

```python
def pascal_triangle(n):
    ans = []
    for i in range(1,n+1):
        res = []
        el = 1
        res.append(el)
        for j in range(1,i):
            el = (el * (i-j))//j
            res.append(el)
        ans.append(res)
    return ans
pascal_triangle(3)
```

```python
def sum_3(a):
    n = len(a)
    ans = []
    for i in range(n):
        for j in range(i+1,n):
            for k in range(j+1,n):
                if (a[i] + a[j] + a[k] == 0):
                    triplet = sorted([a[i],a[j],a[k]])
                    if triplet not in ans:
                         ans.append(triplet)
    return ans
nums = [2, -2, 0, 3, -3, 5]
sum_3(nums)
```

```python
def sum_3_be(a):
    n = len(a)
    ans = []
    for i in range(n):
        for j in range(i+1,n):
            k = -(a[i] + a[j])
            if k in a[j+1:]:
                triplet = sorted([a[i],a[j],k])
                if triplet not in ans:
                    ans.append(triplet)
    return ans
nums = [2, -2, 0, 3, -3, 5]
sum_3_be(nums)
```

```python
def sum_3_has(a):
    n = len(a)
    ans = []
    for i in range(n):
        g = set()
        for j in range(i+1,n):
            k = -(a[i] + a[j])
            if k in g:
                triplet = sorted([a[i],a[j],k])
                if triplet not in ans:
                    ans.append(triplet)
            else:
                g.add(a[j])
    return ans
nums = [2, -2, 0, 3, -3, 5]
sum_3_has(nums)
```

```python
def sum_3_op(a):
    a = sorted(a)
    n = len(a)
    ans = []
    for i in range(n):
        if (i > 0) and a[i] == a[i-1]: 
            continue
        j = i+1
        k = n-1
        while j<k:
            sum = a[i] + a[j] + a[k]
            if sum < 0:
                j+=1
            elif sum > 0:
                k-=1
            else:
                temp = [a[i],a[j],a[k]]
                ans.append(temp)
                j+=1
                k-=1
                while (j < k and a[j] == a[j-1]):
                    j+=1
                while (j < k and a[k] == a[k+1]):
                    k-=1
    return ans

nums = [-1,0,1,2,-1,-4]
sum_3_op(nums)      
```

```python
# brute
def sub_xor(a,k):
    n = len(a)
    count = 0
    for i in range(n):
        for j in range(i,n):
            xor = 0
            for m in range(i,j+1):
                xor = xor ^ a[m]
            if xor == k:
                count+=1
    return count
nums = [4, 2, 2, 6, 4]
k = 6
sub_xor(nums,k)
```

```python
# better
def sub_xo(a,k):
    n = len(a)
    count = 0
    for i in range(n):
        xor = 0
        for j in range(i,n):
            xor = xor ^ a[j]
            if xor == k:
                count+=1
    return count
nums = [4, 2, 2, 6, 4]
k = 6
sub_xo(nums,k)
```

```python
# optimal
def sub_xo_hash(a,k):
    count = 0
    g = {}
    g[0] = 1
    xr = 0
    for i in a:
        xr ^= i
        x = xr ^ k
        count += g.get(x,0)
        g[xr] = g.get(xr,0)+1
    return count
nums = [4, 2, 2, 6, 4]
k = 6
sub_xo_hash(nums,k)
```

```python
def merg_over(a):
    a = sorted(a)
    n = len(a)
    ans = []
    for i in range(n):
        start = a[i][0]
        end = a[i][1]
        if  ans and end <= ans[-1][-1]:
            continue
        for j in range(i,n):
            if a[j][0] <= end:
                end = max(end,a[j][1])
            else:
                break
        ans.append([start,end])
    return ans
c = [[1,3],[2,6],[8,9],[9,11],[8,10],[2,4],[15,18],[16,17]]
merg_over(c)
```

```python
# optimal 
def merge_sub_in(a):
    a = sorted(a)
    n = len(a)
    ans = []
    for i in range(n):
        if not ans or a[i][0] > ans[-1][-1]:
            ans.append(a[i])
        else:
            ans[-1][-1] = max(ans[-1][-1],a[i][1])
    return ans
# c =[[1,3],[2,6],[8,9],[9,11],[8,10],[2,4],[15,18],[16,17]]
c = [[1,4],[4,5]]
merge_sub_in(c)
```

```python
def merge_sor_2(a,b):
    n= len(a)
    m = len(b)
    left1 = 0
    left2 = 0
    ans = []
    while left1 < n and left2 < m:
        if a[left1] < b[left2]:
            ans.append(a[left1])
            left1 +=1
        else:
            ans.append(b[left2])
            left2 +=1
    while left1 < n:
        ans.append(a[left1])
        left1+=1
    while left2 < m:
        ans.append(b[left2])
        left2+=1
    return ans

nums1 = [-5, -2, 4, 5]
nums2 = [-3, 1, 8]
merge_sor_2(nums1,nums2)
```

```python
def merge_sor_s(a,b):
    n = len(a)
    m = len(b)
    left = n-1
    right = 0
    while left >= 0 and right < m:
        if a[left] > b[right]:
            a[left],b[right] = b[right],a[left]
            right +=1
            left -=1
        else:
            break
    a.sort()
    b.sort()
    a.extend(b)
nums1 = [-5, -2, 4, 5]
nums2 = [-3, 1, 8]
merge_sor_s(nums1,nums2)
print(nums1)
```
