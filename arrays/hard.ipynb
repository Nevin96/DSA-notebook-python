```python
def ncr_pascal(n,r):
    res = 1
    for i in range(r):
        res = res*(n-i)
        res = res / (i+1)
    return res

ncr_pascal(3,2)
```

```python
def el_in_row(n):
    ans = []
    for i in range(n):
        c = ncr_pascal(n-1,i)
        ans.append(c)
    return ans
el_in_row(3)
```

```python
def el_in_pas(n):
    ans = []
    res = 1
    ans.append(res)
    for i in range(1,n):
        res = (res * (n - i))//i
        ans.append(res)
    return ans
el_in_pas(3)
```

```python
def pascal_tr(n):
    ans = []
    for i in range(1,n+1):
        l = []
        for j in range(i):
            c = ncr_pascal(i-1,j)
            l.append(c)
        ans.append(l)
    return ans
pascal_tr(3)
```

```python
def pascal_triangle(n):
    ans = []
    for i in range(1,n+1):
        res = []
        el = 1
        res.append(el)
        for j in range(1,i):
            el = (el * (i-j))//j
            res.append(el)
        ans.append(res)
    return ans
pascal_triangle(3)
```

```python
def maj_el(a):
    g = {}
    ans = []
    for i in a:
        if i in g:
            g[i] +=1
        else:
            g[i] = 1
    for key,val in g.items():
        if val > len(a)/3:
            ans.append(key)
    return ans
# nums = [1, 2, 1, 1, 3, 2]
nums = [1, 2, 1, 1, 3, 2, 2]
maj_el(nums)
```

```python
def maj_el(a):
    count1 = 0
    count2 = 0
    el1 = float('-inf')
    el2 = float('-inf')
    for i in a:
        if count1 == 0 and el2 != i:
            el1 = i
            count1+=1
        elif count2 == 0 and el1 != i:
            el2 = i
            count2 +=1
        elif i == el1:
            count1 +=1
        elif i == el2:
            count2 +=1
        else:
            count1 -=1
            count2 -=1
    count1 = 0
    count2 = 0
    ls = []
    for i in a:
        if i == el1:
            count1 +=1
        elif i == el2:
            count2 +=1
    n = len(a)
    if count1 >= (n/3):
        ls.append(el1)
    if count2 >= (n/3):
        ls.append(el2)
    return ls
    
# nums = [1, 2, 1, 1, 3, 2]
# nums = [1, 2, 1, 1, 3, 2, 2]
nums = [1,2]
maj_el(nums)
```
