```python
def l_sub(a,k):
    n = len(a)
    le = 0
    for i in range(n):
        total = a[i]
        for j in range(i+1,n):
            total += a[j]
            if total == k:
                le = max(le,j-i+1)
                break
    return le
nums =  [10, 5, 2, 7, 1, 9]
k=15
l_sub(nums,k)
```

```python
def l_sub_arr(a,k):
    g = {}
    presum =0
    maxlen = 0
    for i in range(len(a)):
        presum += a[i]
        if presum == k:
            maxlen = i+1
        if (presum-k) in g:
            maxlen = max(maxlen,i - g[presum-k])
        if presum not in g:
            g[presum] = i
    return maxlen
nums =  [10, 5, 2, 7, 1, 9]
k=15
l_sub_arr(nums,k)
```

```python
def sum_2(a,k):
    g = {}
    for i in range(len(a)):
        comp = k - a[i]
        if comp in g:
            ind = g[comp]
            return [i,ind]
        if a[i] not in g:
            g[a[i]] = i
nums = [1, 6, 2, 10, 3]
target = 7
sum_2(nums,target)
```

```python
def sort_0_1_2(a):
    n = len(a)
    low = 0
    mid = 0
    high = n-1
    while mid < high:
        if a[mid] == 0:
            a[low],a[mid] = a[mid],a[low]
            low+=1
            mid +=1
        elif a[mid] == 1:
            mid +=1
        elif a[mid] == 2:
            a[mid],a[high] = a[high],a[mid]
            high -=1
    return a
nums = [1, 0, 2, 1, 0]
sort_0_1_2(nums)
```

```python
def maj(a):
    g = {}
    n = len(a)
    for i in a:
        if i in g:
            g[i]+=1
        else:
            g[i] = 1
    for k,v in g.items():
        if v > n/2:
            return k
        
nums = [1, 1, 1, 2, 1, 2]
maj(nums)
```

```python
def rearange_sign(a):
    n = len(a)
    pos = []
    neg = []
    for i in a:
        if i > 0:
            pos.append(i)
        else:
            neg.append(i)
    for i in range(n//2):
        a[2*i] = pos[i]
    for j in range(n//2):
        a[2*j+1] = neg[j]
    return a

nums = [2, 4, 5, -1, -3, -4]
rearange_sign(nums)
```

```python
# optimal
def alt_pos_neg(a):
    pos = 0
    neg = 1
    ar = [0] *len(a)
    for i in a:
        if i > 0:
            ar[pos] = i
            pos+=2
        else:
            ar[neg] = i
            neg+=2
    return ar
nums = [2, 4, 5, -1, -3, -4]
alt_pos_neg(nums)
```

```python
def nxt_per(a):
    n = len(a)
    def rev(start,stop):
        while start < stop:
            a[start],a[stop] = a[stop],a[start]
            start +=1
            stop-=1
    for i in range(n-2,-1,-1):
        if  a[i] < a[i+1]:
            ind = i
            break
    for j in range(n-1,ind+1,-1):
        if a[j] > a[ind]:
            a[ind],a[j] = a[j],a[ind]
            break
    rev(ind+1,n-1)
    return a
        
b = [5,4,7,5,3,2]
nxt_per(b)
```

```python
def lead(a):
    n = len(a)
    b = []
    maxi = float('-inf')
    for i in range(n-1,-1,-1):
        if a[i] > maxi:
            maxi = a[i]
            b.append(maxi)
    return b[::-1]

nums = [1, 2, 5, 3, 1, 2]
lead(nums)
```

```python
def long_cons(a):
    n = len(a)
    count_max = float("-inf")
    for i in range(n):
        count = 1
        while a[i] + count in a:
                count +=1
        else:
            count_max = max(count_max,count)
            
    return count_max
nums = [100, 4, 200, 1, 3, 2]
# nums = [0, 3, 7, 2, 5, 8, 4, 6, 0, 1]
long_cons(nums)
```

```python
def rot_matrix(a):
    n = len(a)
    b = [[0 ]* n for _ in range(n)] 
    for i in range(n):
        for j in range(n):
            b[j][n-1-i] = a[i][j]
    return b

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
rot_matrix(matrix)
```

```python
def tr(a):
    n = len(a)
    for i in range(n-1):
        for j in range(i+1,n):
            a[i][j],a[j][i] = a[j][i],a[i][j]
    for i in range(n):
        a[i].reverse()
    return a
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
tr(matrix)
```

```python
def spir_matr(a):
    n = len(a)
    m = len(a[0])
    top,bottom,right,left = 0,n-1,m-1,0
    ans = []
    while top <= bottom and left <=right:
        for i in range(left,right+1):
            ans.append(a[top][i])
        top +=1
        for i in range(top,bottom +1):
            ans.append(a[i][right])
        right -=1
        if top <= bottom:
            for i in range(right,left-1,-1):
                ans.append(a[bottom][i])
            bottom -=1
        if left <=right:
            for i in range(bottom,top-1,-1):
                ans.append(a[i][left])
            left+=1
    return ans

matrix = [[1, 2, 3], [4 ,5 ,6], [7, 8, 9]]
spir_matr(matrix)
```
