```python
def l_sub(a,k):
    n = len(a)
    le = 0
    for i in range(n):
        total = a[i]
        for j in range(i+1,n):
            total += a[j]
            if total == k:
                le = max(le,j-i+1)
                break
    return le
nums =  [10, 5, 2, 7, 1, 9]
k=15
l_sub(nums,k)
```

```python
def l_sub_arr(a,k):
    g = {}
    presum =0
    maxlen = 0
    for i in range(len(a)):
        presum += a[i]
        if presum == k:
            maxlen = i+1
        if (presum-k) in g:
            maxlen = max(maxlen,i - g[presum-k])
        if presum not in g:
            g[presum] = i
    return maxlen
nums =  [10, 5, 2, 7, 1, 9]
k=15
l_sub_arr(nums,k)
```

```python
def sum_2(a,k):
    g = {}
    for i in range(len(a)):
        comp = k - a[i]
        if comp in g:
            ind = g[comp]
            return [i,ind]
        if a[i] not in g:
            g[a[i]] = i
nums = [1, 6, 2, 10, 3]
target = 7
sum_2(nums,target)
```

```python
def sort_0_1_2(a):
    n = len(a)
    low = 0
    mid = 0
    high = n-1
    while mid < high:
        if a[mid] == 0:
            a[low],a[mid] = a[mid],a[low]
            low+=1
            mid +=1
        elif a[mid] == 1:
            mid +=1
        elif a[mid] == 2:
            a[mid],a[high] = a[high],a[mid]
            high -=1
    return a
nums = [1, 0, 2, 1, 0]
sort_0_1_2(nums)
```

```python
def maj(a):
    g = {}
    n = len(a)
    for i in a:
        if i in g:
            g[i]+=1
        else:
            g[i] = 1
    for k,v in g.items():
        if v > n/2:
            return k
        
nums = [1, 1, 1, 2, 1, 2]
maj(nums)
```

```python
def maj(a):
    count = 0
    for i in a:
        if count == 0:
            candidate = i
            count+=1
        elif i == candidate:
            count+=1
        else:
            count-=1
    return candidate 
nums = [7, 0,0]
maj(nums)
```
