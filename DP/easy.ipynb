```python
def fib(n,dp = None):
    if dp is None:
        dp = [-1] * (n+1)
    if n <= 1:
        return n
    if dp[n] != -1:
        return dp[n]
    dp[n] = fib(n-1,dp) + fib(n-2,dp)
    return dp[n]
fib(4)
```

```python
def f(n):
    dp = [0] * (n+1)
    dp[1] = 1
    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
f(4)
```

```python
def f(n):
    prev = 1
    prev2 = 0
    for i in range(2,n+1):
        curr = prev + prev2
        prev2,prev = prev,curr
    return prev
f(4)
```

```python
def cl_step(n):
    if n == 0:
        return 1
    if n== 1:
        return 1
    left = cl_step(n-1)
    right = cl_step(n-2)
    return left + right
cl_step(2)
```

```python
def cl(n):
    dp = [1] * (n+1)
    dp[0] = 1
    dp[1] = 1
    for i in range(2,n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
cl(3)
```

```python
def climb(n):
    if n <= 1:
        return 1
    st1 = 1
    st2 = 1
    for _ in range(2,n+1):
        curr = st1 + st2
        st2  = st1
        st1 = curr
    return curr
climb(3)
```

```python
def fr(a):
    n = len(a)
    dp = [0] * n
    for i in range(1,n):
        fs = dp[i-1] + abs(a[i] - a[i-1])
        ss = float('inf')
        if i > 1:
            ss = dp[i-2] + abs(a[i] - a[i-2])
        dp[i] = min(fs,ss)
    return dp[n-1]
heights = [2, 1, 3, 5, 4]
fr(heights)
```

```python
def frog(a,k,ind= None,dp = None):
    n = len(a)
    if dp == None:
        dp = [-1] * n
    if ind is None:
        ind = n-1
    if ind == 0:
        return 0
    if dp[ind] != -1:
        return dp[ind]
    mini = float('inf')
    for i in range(1,k+1):
        prev = ind - i
        if prev < 0:
            break
        c = frog(a,k,prev,dp) + abs(a[ind] - a[prev])
        mini = min(c,mini)
    dp[ind] = mini
    return mini
heights = [15, 4, 1, 14, 15]
k = 3
frog(heights,k)
```
