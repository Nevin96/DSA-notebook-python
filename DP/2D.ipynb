```python
def unique_path(m,n):
    def path(i,j):
        if i == j == 0:
            return 1
        elif i < 0 or j < 0 or i==m or j ==n:
            return 0
        else:
            return path(i,j-1) + path(i-1,j)
    return path(m-1,n-1)

unique_path(3,7)
```

```python
def minfalling_path(matrix): 
    m = len(matrix)
    n = len(matrix[0]) 
    dp = [[-1] * n for i in range(m)]
    def minfalling(matrix,i,j,dp):
        m = len(matrix)
        n = len(matrix[0])
        if j < 0 or j >= n:
            return float('inf')
        if i == 0:
            dp[0][j] = matrix[0][j]
            return matrix[0][j]
        if dp[i][j] != -1:
            return dp[i][j]
        up = minfalling(matrix,i-1,j,dp) + matrix[i][j]
        l_up = minfalling(matrix,i-1,j-1,dp) + matrix[i][j]
        r_up = minfalling(matrix,i-1,j+1,dp) + matrix[i][j]
        dp[i][j] = min(up,l_up,r_up)
        return dp[i][j]
    ans = float('inf')
    for j in range(n):
        ans = min(ans,minfalling(matrix,m-1,j,dp))
    return ans

    
matrix = [[1, 2, 10, 4], [100, 3, 2, 1], [1, 1, 20, 2], [1, 2, 2, 1]]
minfalling_path(matrix)
```

```python
def minfall(matrix):
    m = len(matrix)
    n = len(matrix[0])
    dp = [[-1] * n for _ in range(m)]
    for i in range(n):
        dp[0][i] = matrix[0][i]
    for i in range(1,m):
        for j in range(n):
            ans = float('inf')
            up = matrix[i][j] + dp[i-1][j]
            diar = float('inf')
            dial = float('inf')
            if j > 0:
                dial = matrix[i][j] + dp[i-1][j-1]
            if j < n-1:
                diar = matrix[i][j] + dp[i-1][j+1]
            ans = min(up,diar,dial)
            dp[i][j] = ans
    return min(dp[m-1])
    
# matrix = [[1, 2, 10, 4], [100, 3, 2, 1], [1, 1, 20, 2], [1, 2, 2, 1]]
matrix = [[1, 4, 3, 1], [2, 3, -1, -1], [1, 1, -1, 8]]
minfall(matrix)
```
