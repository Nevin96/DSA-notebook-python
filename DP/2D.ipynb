```python
def unique_path(m,n):
    def path(i,j):
        if i == j == 0:
            return 1
        elif i < 0 or j < 0 or i==m or j ==n:
            return 0
        else:
            return path(i,j-1) + path(i-1,j)
    return path(m-1,n-1)

unique_path(3,7)
```

```python
def uniw(m,n):
    memo = {(0,0): 1}
    def path(i,j):
        if (i,j) in memo:
            return memo[(i,j)]
        elif i < 0 or j < 0 or i >= m or j >= n:
            return 0
        else:
            val = path(i,j-1) + path(i-1,j)
            memo[(i,j)] = val
        return val
    return path(m-1,n-1)
unique_path(3,7)
```

```python
def u(m,n):
    dp = []
    for _ in range(m):
        dp.append([0] * n)
    dp[0][0] = 1
    for i in range(m):
        for j in range(n):
            if i == j == 0:
                continue
            val = 0
            if i > 0:
                val += dp[i-1][j]
            if j > 0:
                val += dp[i][j-1]
            dp[i][j] = val
    return dp[m-1][n-1]
u(3,7)
```

```python
def uni_2(grid):
    m = len(grid)
    n = len(grid[0])
    if grid[0][0]:
        grid[0][0] = 0
    else:
        grid[0][0] = 1
    for  i in range(m):
        for j in range(n):
            if i == j == 0:
                continue
            elif grid[i][j]:
                grid[i][j] = 0
            elif i > 0 and j > 0:
                grid[i][j] = grid[i][j-1] + grid[i-1][j]
            elif i > 0:
                grid[i][j] = grid[i-1][j]
            elif j > 0:
                grid[i][j] = grid[i][j-1]
    return grid[m-1][n-1]
obstacleGrid = [[0,0,0],[0,1,0],[0,0,0]]
uni_2(obstacleGrid)
```

```python
def nin(tasks,last,day = None):
    m = len(tasks)
    n = len(tasks[0])
    if day is None:
        day = m-1
    if day == 0:
        maxi = float('-inf')
        for i in range(n):
            if i != last:
                maxi = max(maxi,tasks[day][i])
        return maxi
    mx = float('-inf')
    for i in range(n):
        if i != last:
            points = nin(tasks,i,day-1) + tasks[day][i]
            mx = max(points,mx)
    return mx
a =[[10, 40, 70], [20, 50, 80], [30, 60, 90]]
nin(a,3)
```

```python
def nin(tasks,last,day = None,dp = None):
    m = len(tasks)
    n = len(tasks[0])
    if day == None:
        day = m-1
    if dp == None:
        dp = [[-1] * (n+1) for _ in range(m)]
    if dp[day][last] != -1:
        return dp[day][last]
    if day == 0:
        maxi = float('-inf')
        for i in range(n):
            if i != last:
                maxi = max(maxi,tasks[day][i])
        dp[day][last] = maxi
        return maxi
    mx = float('-inf')
    for i in range(n):
        if i != last:
            points = tasks[day][i] + nin(tasks,i,day-1,dp)
            mx = max(mx,points)
    dp[day][last] = mx
    return mx
a =[[10, 40, 70], [20, 50, 80], [30, 60, 90]]
nin(a,3)
```

```python
def ni(tasks):
    m = len(tasks)
    n = len(tasks[0])
    dp = [[-1] * (n+1) for _ in range(m)]

    for last in range(n+1):
        maxi = 0
        for i in range(n):
            if i != last:
                maxi = max(tasks[0][i],maxi)
        dp[0][last] = maxi
    for day in range(1,m):
        for last in range(n+1):
            maxi = 0
            for task in range(n):
                if task != last:
                    points = dp[day-1][task] + tasks[day][task]
                    maxi = max(points,maxi)
            dp[day][last] = maxi
    return dp[m-1][n]
a =[[10, 40, 70], [20, 50, 80], [30, 60, 90]]
ni(a)
```

```python
def ninj(tasks):
    m = len(tasks)
    n = len(tasks[0])
    dp = [-1] * (n+1) 
    for last in range(n+1):
        maxi = 0
        for i in range(n):
            if i != last:
                maxi = max(tasks[0][i],maxi)
        dp[last] = maxi
    for days in range(1,m):
        temp = [-1] * (n+1)
        for last in range(n+1):
            maxi = 0
            for i in range(n):
                if i != last:
                    point = dp[i] + tasks[days][i]
                    maxi = max(point,maxi)
            temp[last] = maxi
        dp = temp
    return dp[n]
a =[[10, 40, 70], [20, 50, 80], [30, 60, 90]]
ninj(a)
```
